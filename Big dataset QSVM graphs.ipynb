```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix
```

```python
out_dir = r"C:\Users\Ssanp\OneDrive\Documents\SRP 2026"

classical_results = pd.read_csv(out_dir + r"\big_classical_results.csv")
qsvm_results = pd.read_csv(out_dir + r"\big_qsvm_results.csv")

results = pd.concat([classical_results, qsvm_results], ignore_index=True)

results
```

```python
metrics = ["Accuracy", "Precision", "Recall", "F1 Score"]

results.set_index("Model")[metrics].plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Big SAR Classical vs QSVM Performance")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.legend(title="Metric")
plt.tight_layout()
plt.show()
```
